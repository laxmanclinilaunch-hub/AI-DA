# -Break , continue , pass  

- Break:

when specific condition is met ,break allows to teriminate the loop

```python
y="python"


for i in y:
    print(i)
```

```python
y="python"


for i in y:
    if i=="t":
        break
    print(i)
```

```python
z="seema is learning python"


for i in z:
    if i=="e":
        break
    print(i)
```

# -continue

when specific condition is met , continue allows to skip that part  and control transfer to next statement 

```python
z="seema is learning python"


for i in z:
    if i=="e":
        continue
    print(i)
```
