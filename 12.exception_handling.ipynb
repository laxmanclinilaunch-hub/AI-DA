```python
def divide(num):
    if num==0:
        raise Exception ("can't divide with zero")
    else:
        r=12/num
    return r    
```

```python
try:
    print(divide(6))

except Exception as e:
    print(e)
```

```python
try:
    print(divide(0))

except Exception as e:
    print(e)
```
