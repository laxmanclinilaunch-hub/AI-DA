- class method:

   - which can be defined with using the decorator @classmethod and along with you need to use cls parameter and using class variables 

```python
class netflix:
    new_show="kalki"


    def __init__(self,mail,password):
        self.mail=mail
        self.password=password
        self.service="Free"
        self.amount=0


    def get_password(self):
        return self.password


    @classmethod
    def get_newshow(cls):
        return cls.new_show


    @classmethod
    def change_newshow(cls,show_name):
        cls.new_show=show_name
```

# - encapsulation:

- it is  a pillar of oops
- it using varibles and methods we have to controlling access to the data

-  a core concept in programming where data and the functions that use that data are bundled together into a single unit





- public memeber:

  - it can access anywhere (inside the class or outside of the class)

           - self.name=name



- proteced member:


-  the protected memeber start with one underscore
 - it can access inside the class and its subclass ( child class)

        - self._name = name 

- private memeber
     - private memeber  start with duble underscore

     - the private memeber can't access directly 

     - to access private member we have to (getter)  method and for modification use (setter) method 

                   - self.__name = name 





- public member

```python
class mentor:
    def __init__(self):
        self.name="laxman"
        


obj=mentor()

print(obj.name)
```

- protected member

```python
class mentor:
    def __init__(self):
        self._name='laxman'
        

obj=mentor()

print(obj._name)  # protected member

```

```python
class father:
    def __init__(self):
        self._salary=70000




class son(father):
    def display(self):
        print(self._salary)



obj=son()


obj.display()
```

- for accessing of private member we have to use getter method

```python
class mentor:
    def __init__(self):
        self.__name="laxman"


    def get_method(self):
            return self.__name


obj=mentor()

print(obj.get_method())
```

```python
class mentor:
    def __init__(self):
            self.__name="laxman"

    def setter_method(self,change):
          self.__name=change


    def get_method(self):
          return  self.__name   


obj=mentor()


obj.setter_method("arunachalam")

print(obj.get_method())
```

# Abstraction

- it is the process of hiding the internal implementation details and exposing only the essential functionality to the user

- generally for impplementation of this using ABC and @abstractmethod

```python
from abc import ABC,abstractmethod


class atm(ABC):
    @abstractmethod
    def withdraw_money(self, amount):
        pass


class sbi_atm(atm):

    def withdraw_money(self, amount):
        print(f"verifying account...")
        print(f"processsing of withdrawl") 
        print(f" ${amount} withdraw successfully")   



obj=sbi_atm()

obj.withdraw_money(50000)

```

# inheritance :
- it acquring the properties from parents class(Base class) to child class ( derived class)

- what is properties?

behaviour from one class to anothers i.e means it's attributes ( variables) and methods ( functions)

- single inheritance:
   - transfer the properties from single parent to single child

```python
class person():

    def __init__(self,name,idnumber):
        self.name=name
        self.idnumber=idnumber


    def details(self):
        print("my name is",self.name)
        print("my id number is ",self.idnumber) 

class employee(person):

    def __init__(self, name, idnumber,salary,post):
        person.__init__(self,name,idnumber)
                # or 
        # super().__init__(name,idnumber)
        self.salary=salary
        self.post=post


    def show_details(self):
        print("my name is",self.name)
        print("my id number is ",self.idnumber)
        print("my salary is ",self.salary)


# creating obj

obj=employee("pratik",6879,90000,"dentist")


obj.details()

print()

obj.show_details()        
```

- multiple inheritance




```python
class f:
    def __init__(self,id):
        self.id=id


    def details(self):
        print("this is my id number",self.id)



class m:
    def __init__(self,name):
        self.name=name


    def show(self):
        print("my name is ",self.name)



class child(f,m):

    def __init__(self, id,name,salary):
        f.__init__(self,id)
        m.__init__(self,name)

        # or 
        # super().__init__(id,name)
        self.salary=salary

    def employee_details(self):
        print('this is id number',self.id)
        print("my name is ",self.name)
        print("this is my salary",self.salary)



# creating obj
obj=child(34567,"pratik",55000)

obj.employee_details()

print()

obj.details()

print()

obj.show()


```

- multilevel inheritance:
 

 

```python
class grandparent():
    def __init__(self,grandparent_name):
        self.grandparent_name=grandparent_name

    def display_grandparents(self):
        return "Grand_parent:",self.grandparent_name


class parent(grandparent):
    def __init__(self, grandparent_name,parent_name):
        super().__init__(grandparent_name)   
        self.parent_name=parent_name

    def display_parent(self):
        return "parent_name:",self.parent_name


class child(parent):
    def __init__(self, grandparent_name, parent_name,child_name):
        super().__init__(grandparent_name, parent_name)   
        self.child_name=child_name

    def display_child(self):
        return "child_name:",self.child_name

# creating obj

obj=child("Babu","bhvana","manisha")


print(obj.display_grandparents())

print(obj.display_parent())

obj.display_child()

         
        
```

# Hierarchical inheritance:

# Hybrid inheritance:

- Hybrid inheritance in Python is a combination of two or more types of inheritance,

- such as combining hierarchical inheritance with multiple inheritance

- This structure often forms what is known as the "diamond problem"

```python
class school():
    def fun1(self):
        print("this is school function")



class student1(school):
    def fun2(self):
        print("this is student1 function")



class student2(school):
    def fun3(self):
        print("this is student2 function")


class student3(student1,school):
    def fun4(self):
        print("this is student3 fundction") 



# creating obj
obj=student3()


obj.fun1()
print()
obj.fun2()
print()
obj.fun4()
```
