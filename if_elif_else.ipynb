# If- elif - else:

- if statements in python allows us to tell the computer to perform alternative actions based on certain set conditional results

- we are telling to the computer "Hey if this case happens , perform someaction, otherwise if another case happens perform someother action , otherwise if all other actions are failed perform this action only"

if case1:

     perform action1
elif case2:

     perofmr action2
else:

    perform action3
