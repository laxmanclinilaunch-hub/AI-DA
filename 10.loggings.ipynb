# loggings:

it is a crucial aspect of any application


- it is used to track the events , erros, any other operational information

- python built in -logging modules offers a flexibale framework for getting log messages

```python
import logging

import os
```

```python
log_dir="loogings"

log_file="app.log"


os.makedirs(log_dir,exist_ok=True)

log_path=os.path.join(log_dir,log_file)
```

```python
logging.basicConfig(filename=log_path,level=logging.INFO)
```

- LOG LEVELS

- 1.INFO:


confirmation that things are working expected 

```python
logging.info("this is my logs of info message")
```

- 2. warning:

An indication that something unexpected happend or indicating of some problem in the near future


example :
 low level disk space

```python
logging.warning("this is my logs of warning message")
```

- Error:


Due to some serious problem , the software or system has unable to perform the actions

```python
logging.error("this is my logs of error message")
```

# critial log:

- it is very seriuos problem, if effect the continues running of some software action

```python
logging.critical("this is my log of  crtical message error")
```

# Debug:

- detailed information, it works when get problem

```python
logging.debug("this is my logs of debug error")
```

- Advance logs:


```python
log_dir="loogings"

log_file="app.log"


os.makedirs(log_dir,exist_ok=True)

log_path=os.path.join(log_dir,log_file)
```

```python
logging.basicConfig(filename=log_path,level=logging.INFO,format="[%(asctime)s] %(name)s -%(levelname)s - %(message)s",force=True)
```

```python
logging.info("this is my logs of info message")
```

```python
logging.warning("this is my logs of warning message")
```

```python
logging.error("this is my logs of error message")
```

```python
def add(a,b):
    result = a + b
    logging.info(f"Adding {a} and {b}, result: {result}")
    return result

def subtract(a,b):
    result = a - b
    
    return logging.info(f"Subtracting {b} from {a}, result: {result}")


def multiply(a,b):
    result = a * b
    
    return logging.info(f"Multiplying {a} and {b}, result: {result}")  

def divide(a,b):
    if b == 0:
        logging.error("Attempted to divide by zero.")
        raise ValueError("Denominator cannot be zero.")
    result = a / b
    
    return logging.info(f"Dividing {a} by {b}, result: {result}")
```

```python
add(5,2)
```

```python
subtract(5,2)
```

```python
multiply(5,2)
```

```python
divide(10,2)
```
