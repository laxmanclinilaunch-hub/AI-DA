```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
healthcare = pd.read_csv('healthcare.csv')
```

```python
healthcare.head()
```

```python
admissions = pd.read_csv(r'admissions.csv')
```

```python
admissions.head()
```

## 1. Scatterplot (Numerical - Numerical)

```python
sns.scatterplot(x=healthcare['BMI'], y=healthcare['BloodPressure'],hue=healthcare["Gender"] ,style=healthcare['Smoker'])
```

```python
sns.scatterplot(x=healthcare['BMI'], y=healthcare['BloodPressure'],
                hue=healthcare['Gender'], style=healthcare['Smoker'], size=healthcare['InsurancePlan'])
```

## 2. Bar Plot (Numerical - Categorical)

```python
healthcare.head()
```

```python
sns.barplot(x=healthcare['Gender'], y=healthcare['Age'], hue=healthcare['InsurancePlan'])
```

## 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(x=healthcare['Gender'], y=healthcare['Age'], hue=healthcare['HeartDisease'], palette=["red", "blue"])
```

## 4. Distplot (Numerical - Categorical)

```python
sns.histplot(healthcare[healthcare['HeartDisease']==0]['Age'], kde=True, stat='density', label='No Heart Disease')
sns.histplot(healthcare[healthcare['HeartDisease']==1]['Age'], kde=True, stat='density', label='Heart Disease')
plt.legend()
```

## 5. HeatMap (Num- Num)

```python
healthcare.head(3)
```

```python
healthcare.corr(numeric_only=True)
```

```python
healthcare.drop(columns=["PatientId"],inplace=True)
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(healthcare.corr(numeric_only=True),fmt=".2f",annot=True)
```

## 6. ClusterMap (num-num)

```python
sns.clustermap(pd.crosstab(healthcare['NumDependents'], healthcare['HeartDisease']))
```

## 7. Pairplot

```python
vitals = healthcare[['BMI', 'BloodPressure', 'Cholesterol', 'Glucose', 'Diagnosis']]
vitals.head()
```

```python
sns.pairplot(vitals, hue='Diagnosis')
```
