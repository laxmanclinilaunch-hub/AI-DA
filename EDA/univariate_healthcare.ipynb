- Q-Q plot 

```python
import scipy.stats as stats

stats.probplot(df['Glucose'], dist="norm", plot=plt)

plt.title('Q-Q Plot - Glucose')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('Sample Quantiles')
plt.show()
```
