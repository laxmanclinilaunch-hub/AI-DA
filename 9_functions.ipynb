```python
def calculate_area(l,w):
    """
    calculate area of rectangle
    parameters:
    length (l)[float]: the length of rectangle
    width (w)[float]: the width f rectangle

    return:
    float :  the area of the rectangle 
    """
    return l*w

def calculate_perimeter(l,w):
    """
        calculate perimeter of rectangle
        parameters:
        length (l)[float]: the length of rectangle
        width (w)[float]: the width f rectangle
    
        return:
        float :  the perimeter of the rectangle 
        """
    return 2*(l+w)


def print_shape(l,w):
    """
    it prints area & perimeter of rectangle
    parameters:
            length (l)[float]: the length of rectangle
            width (w)[float]: the width of rectangle
    returns:  area & perimeter of rectangle
    """

    area=calculate_area(l,w)
    perimeter=calculate_perimeter(l,w)

    print(f"Rectangle with length {l} and width {w}:")

    print(f"Area {area}")

    print(f"Perimeter {perimeter}")
```

```python
print_shape(25.0,10.6)
```

- LEGB Rule:
first it checks local (inside the fun) ,next check enclosing (inside any enclosing func(nested fun)) then afterthat global variavle (outside the function) , Built-in  finally it cheks the built in scope

```python
def outer_fun():
    x=7
    def inner_fun():
        x=5
        print(x)

    inner_fun()
    print(x)  # This will print the value of x in the outer function



# calling of my function
outer_fun()        
```
