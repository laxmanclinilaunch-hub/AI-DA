- Docstring for the function

```python
def odd_even(n:int):
    """
    Determines if a number is odd or even.
    
    Parameters:
    n (int): The number to check.

    outputs:
    the ouput will be printed as odd or even

    created date; 2024-06-20

    created by: Lakshman
    """

    if type(n)==int:
        if n%2==0:
            print(f"{n} is even")
        else:
            print(f"{n} is odd")


    else:
        print("Please enter a valid integer")        
```

```python
def calculate_area(l,w):
    """
    calculate area of rectangle
    parameters:
    length (l)[float]: the length of rectangle
    width (w)[float]: the width f rectangle

    return:
    float :  the area of the rectangle 
    """
    return l*w

def calculate_perimeter(l,w):
    """
        calculate perimeter of rectangle
        parameters:
        length (l)[float]: the length of rectangle
        width (w)[float]: the width f rectangle
    
        return:
        float :  the perimeter of the rectangle 
        """
    return 2*(l+w)


def print_shape(l,w):
    """
    it prints area & perimeter of rectangle
    parameters:
            length (l)[float]: the length of rectangle
            width (w)[float]: the width of rectangle
    returns:  area & perimeter of rectangle
    """

    area=calculate_area(l,w)
    perimeter=calculate_perimeter(l,w)

    print(f"Rectangle with length {l} and width {w}:")

    print(f"Area {area}")

    print(f"Perimeter {perimeter}")
```

```python
print_shape(25.0,10.6)
```

```python
def cal_area(length,width,height,thickness):
    return length*width*height*thickness
```

```python
cal_area(10,5,2,3)
```

```python
def cal_area(*args):
    product=1
    for i in args:
        product=product * i
        
    return product
```

```python
cal_area(10,5)
```

```python
cal_area(10,5,2)
```

```python
cal_area(10,5,2,3)
```

```python
cal_area(10,5,2,9,5,2,6)
```
